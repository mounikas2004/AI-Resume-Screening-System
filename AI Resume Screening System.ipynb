```python
libraries imported

import pandas as pd
import numpy as np

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print("All libraries imported successfully!")
```

```python
Job Description data create

resume = """
I am a Computer Science graduate with skills in Python,
Machine Learning, SQL, Pandas, NumPy and Data Analysis.
I have experience in developing machine learning projects
and working with data preprocessing and visualization.
"""
job_description = """
We are looking for a Python Developer with knowledge of
Python, Machine Learning, SQL, Pandas, NumPy and Data Analysis.
The candidate should have experience in developing
machine learning applications.
"""
print("Resume:")
print(resume)

print("\nJob Description:")
print(job_description)
```

```python
AI Match Score calculate


vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform([
    resume,
    job_description
])
similarity = cosine_similarity(
    tfidf_matrix[0:1],
    tfidf_matrix[1:2]
)
match_score = similarity[0][0] * 100
print("==============================")
print("   AI RESUME SCREENING SYSTEM")
print("==============================")
print(f"Resume Match Score: {match_score:.2f}%")
```

```python
Matched Skills ಮತ್ತು Missing Skills


skills = [
    "Python",
    "Machine Learning",
    "SQL",
    "Pandas",
    "NumPy",
    "Data Analysis",
    "Data Visualization",
    "Deep Learning",
    "Java",
    "C++",
    "Django",
    "Flask"
]
resume_lower = resume.lower()
job_lower = job_description.lower()
matched_skills = []
missing_skills = []
for skill in skills:
    if skill.lower() in job_lower:
        if skill.lower() in resume_lower:
            matched_skills.append(skill)
        else:
            missing_skills.append(skill)

print("================================")
print("       SKILL ANALYSIS")
print("================================")

print("\nMatched Skills:")
for skill in matched_skills:
    print("✓", skill)

print("\nMissing Skills:")
for skill in missing_skills:
    print("✗", skill)
```

```python
Shortlist Recommendation

if match_score >= 70:
    recommendation = "SHORTLISTED"
elif match_score >= 50:
    recommendation = "REVIEW REQUIRED"
else:
    recommendation = "NOT SHORTLISTED"
print("================================")
print("       FINAL SCREENING RESULT")
print("================================")
print(f"Match Score: {match_score:.2f}%")
print(f"Recommendation: {recommendation}")
```
